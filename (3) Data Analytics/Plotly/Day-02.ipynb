```python
import plotly.graph_objects as go
```

# Non Convex Chart

```python
# Non Convex 3D Chart

# z = sin(x) * cos(y)

import numpy as np

x = np.linspace(-5,5,30)          # x co-ordinate with 30 data points from -5 to 5
y = np.linspace(-5,5,30)          # y co-ordinate with 30 data points from  -5 to 5

x_new, y_new = np.meshgrid(x,y)      # Creates a Grid in which 3D Chart will be plotted

z = np.sin(x_new) * np.cos(y_new)     # z co-ordinate

Fig_2 = go.Figure( data = [go.Surface( z = z, x = x_new, y = y_new )]).update_layout( width = 800, height = 600 )
Fig_2.show()

# update_layout() : Resize the Chart
# go.Surface( z = z, x = x_new, y = y_new) : Here, We are Passing 3 Co-ordinates
```

# 3D Scatter Plot

```python
Sales = np.random.randint(5,20,30)           # Generating 30 Values from 5 to 20
Quality = np.random.randint(5,25,30)
Footfalls = np.random.randint(5,25,30)

Fig_3 = go.Figure(data = go.Scatter3d(x = Sales, y = Quality, z = Footfalls, mode = 'markers'))

Fig_3.update_layout( width = 800,
                     height = 600,
                     title = 'Sales Data',
                     scene = dict(
                         xaxis_title = "Sales",
                         yaxis_title = "Quality",
                         zaxis_title = "Foot falls")
                     )
Fig_3.show()
```

# 3D Multi Scatter Plot

```python
# Scatter Plot - 1

Sales_2 = np.random.randint(5,20,30)           # Generating 30 Values from 5 to 20
Quality_2 = np.random.randint(5,25,30)
Footfalls_2 = np.random.randint(5,25,30)

# Scatter Plot - 2

Sales_3 = np.random.randint(5,20,30)           # Generating 30 Values from 5 to 20
Quality_3 = np.random.randint(5,25,30)
Footfalls_3 = np.random.randint(5,25,30)

Fig_4 = go.Figure()

Fig_4.add_trace(go.Scatter3d(x = Sales_2, y = Quality_2, z = Footfalls_2, mode = 'markers'))
Fig_4.add_trace(go.Scatter3d(x = Sales_3, y = Quality_3, z = Footfalls_3, mode = 'markers'))

Fig_4.update_layout( width = 800,
                     height = 600,
                     title = 'Sales Data',
                     scene = dict(
                         xaxis_title = "Sales",
                         yaxis_title = "Quality",
                         zaxis_title = "Foot falls")
                     )
Fig_4.show()
```
