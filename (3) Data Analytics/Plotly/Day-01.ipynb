# Convex 3D Plot

```python
# z = (x ** 2) + (y ** 2)

import numpy as np
```

```python
x = np.linspace(-5,5,30)
y = np.linspace(-5,5,30)

x_new, y_new = np.meshgrid(x,y)      # Creates a Grid in which 3D Chart will be plotted

z = ( x_new ** 2 ) + ( y_new ** 2 )

Fig = go.Figure( data = [go.Surface( z = z, x = x_new, y = y_new )]).update_layout( width = 800, height = 600 )
Fig.show()
```
