```python
import pandas as pd
```

```python
# Converting a Dictionary to DataFrame

England = { "TEST" : ("Joe Root","Harry Brook","Jofra Archer"),
            "ODI" : ("Joe Root","Harry Brook","Jos Buttler"),
            "T20I" : ("Jos Buttler","Harry Brook","Phil Salt")}

Team = pd.DataFrame(England, index = (1,2,3))
print(Team)
print(type(Team))

# Keys will become Column_Names while converting a Dictionary into DataFrame.
# Values will be inserted in that Column.
```

```python
# Converting a Tuple into DataFrame

Movies = ((1,"Jersey","Gowtham Thinnanuri"),
          (2,"K.G.F","Prashanth Neel"),
          (3,"The Prestige","Christopher Nolan"))

Best_Movies = pd.DataFrame(Movies, columns = ("Rank","Movie Name","Director"), index = (1,2,3))
print(Best_Movies)
print(type(Best_Movies))
```

```python
# Converting a Set into Dataframe

# We Cannot store a Set inside a Set
# We Cannot store the Mutable items inside the set

Fruits = {("Apple",100),
          ("Mango",80),
          ("Banana",20),
          ("Orange",30)}

Fruits_Price = pd.DataFrame(Fruits, columns = ("Fruits","Price"))
print(Fruits_Price)
print(type(Fruits_Price))
```

```python
# We Can Convert list of list into DataFrame
# We Can Convert list of tuple into DataFrame
# We Can Convert tuple of tuple into DataFrame
# We Cannot store Set inside the set
# We Can Convert set of tuple into DataFrame
```

```python
# Converting a Series into DataFrame

Series = pd.Series([10,20,30,40,50,60])

DF = pd.DataFrame(Series, columns = ["Values"])
print(DF)
print(type(DF))
```
