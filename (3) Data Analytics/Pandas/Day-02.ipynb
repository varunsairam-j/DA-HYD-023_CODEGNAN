```python
import pandas as pd
```

```python
# Create the Pandas Series from dictionary
Brook = {"Test" : 317, "ODI" : 201, "T20I" : 114}

Ser_1 = pd.Series(Brook)
print(Ser_1)
print(type(Ser_1))
```

```python
# Now, Create Tuple
Root = ("Test",262,"ODI",166,"T20I",90)

Ser_2 = pd.Series(Root)
print(Ser_2)
print(type(Ser_2))
```

```python
# Creating a Single element pandas Series

Ser_4 = pd.Series(100)
print(Ser_4)
print(type(Ser_4))
```

# Update, delete, insert Operations on Series

```python
Ser_5 = pd.Series([10,20,30,40,50])
print(Ser_5)
```

```python
# Update Operation

Ser_5[3] = 100
print(Ser_5)
```

```python
# Insert Operation

# If the Index is not present, the value will be added at the end of the Series.

Ser_5[5] = 230
print(Ser_5)
```

```python
# We can insert the element with any new index value
# Index value need not to be in sequence

Ser_5[10] = 420
print(Ser_5)
```

```python
# We can have different index types (str or int or anything)

Ser_5['A'] = 1000
print(Ser_5)
```

```python
Ser_5['A'] = 5000
print(Ser_5)
```

```python
# Delete element from Series

Ser_5 = Ser_5.drop(10)
print(Ser_5)
```

```python
# Multiple elements deletion 

Ser_5 = Ser_5.drop([1,3,'A'])
print(Ser_5)
Ser_5 = Ser_5.drop([2,4])
print(Ser_5)
```

# Dataframe : 2D Array

```python
# Example-1 : Creating empty dataframe

df = pd.DataFrame()
print(df)
```

```python
Score = [400,264,175,333,183,200]

Df_1 = pd.DataFrame(Score)
print(Df_1)
print(type(Df_1))
```

```python
England = [("Banton","Duckett","Bethell","Root","Brook"),
           ("Buttler","Jacks","Curran","Overton","Archer"),
           ("Dawson","Rashid","Atkinson","Tongue","Baker")]

Df_2 = pd.DataFrame(England)
print(Df_2)
print(type(Df_2))
```

```python
# Making the Dataframe Meaningful

Players = [("Duckett","Bethell","Buttler","Root","Brook"),
           ("Jacks","Curran","Overton","Dawson","Coles"),
           ("Archer","Rashid","Tongue","Baker","Crocombe")]

Df_3 = pd.DataFrame(Players, columns = [1,2,3,4,5], index = ["Batsmen","All-Rounders","Bowlers"])
print(Df_3)
```
