# Data Filtering and Indexing (loc and iloc)

```python
# Find the reports based on given condition and indexing 
```

```python
# Query  :  Filter all the records where salary is greater than 1,00,000

# Data.salary > 100000   -->   gives True or False
Data[Data.salary > 100000]
```

```python
# Q-2  :  Find the records where salary < 50000

Data[Data.salary < 50000]
```

```python
# Q-3  :  Find the records where salary > 50000 and salary < 150000

Data[(Data.salary > 50000) & (Data.salary < 150000)]
```

```python
# Q-4  :  Find the records where salary < 50000 or salary > 150000

Data[(Data.salary < 50000) | (Data.salary > 150000)]
```

# loc & iloc for indexing 

```python
Data.columns
```

```python
# loc and iloc is used for filtering the records based upon indexing

Data.loc[1:4, 'ID' : 'salary']     # Data.loc[row,column]
```

```python
# Print 5 to 15 rows and Age to salary column using loc 

Data.loc[5:15, 'ID' : 'salary']
```

```python
# Print 1 to 20 rows and Name to salary column using loc 

Data.loc[1:20:2, 'Name' : 'salary']
```

```python
# Use of iloc

Data.iloc[1:4,0:4]

# printing records from index 1 to 4-1 & columns from 0th to 4-1th column
```

```python
# print reports from 4 to 14 and columns from 2 to 4

Data.iloc[4:15,2:5]
```

```python
# print even no of records from 1 to 10 and odd no of columns from 1 to 5

Data.iloc[2:11:2,1:6:2]     # iloc[row(start,end,step), col(start,end,step)]
```

```python
# use loc and print records from 1 to 10 and 2 specific columns 'Name' and 'salary'

Data.loc[1:10, 'Name' : 'salary':3]     # or  Data.loc[1:10, ['Name','salary']] 
```

```python
# use iloc and print records from 1 to 10 and 2 specific columns 'Name' and 'salary'

Data.iloc[1:11, 1:5:3]     #  Data.iloc[1:11,[1,4]]
```

```python
# Condition inside the loc and iloc

# Query  :  Print Name of People having Age > 50

Data.loc[Data.Age > 50, ["Name","Age"]]
```

```python
# using iloc

Data.iloc[Data.Age > 50, [1,2]]

# If older version, Data.iloc[(Data.Age > 50).values, [1,2]]
```
