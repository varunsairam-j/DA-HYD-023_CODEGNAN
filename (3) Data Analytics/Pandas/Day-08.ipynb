# Transformation using map()

```python
# Map()  :  This function is mainly used for values mapping

# Mapping one value to another

import pandas as pd
Data = pd.read_excel("Excel_Data.xlsx")
Data
```

```python
Data.Profession.unique()
```

```python
Data.Profession = Data.Profession.str.strip()
Data.Profession.unique()
```

```python
Profession_map_dict = { 
    'Teacher' : 'ఉపాధ్యాయుడు',
    'Boxer' : 'బాక్సర్',
    'Scientist' : 'శాస్త్రవేత్త',
    'Chef' : 'చెఫ్',
    'Doctor' : 'వైద్యుడు',
    'Artist' : 'కళాకారుడు',
    'Comedian' :  'హాస్యనటుడు',
    'Software Engineer' : 'సాఫ్ట్‌వేర్ ఇంజనీర్',
    'Detective' : 'డిటెక్టివ్',
    'Pilot' : 'పైలట్'             
}

Data['mapped_profession'] = Data.Profession.map(Profession_map_dict)
Data
```

```python
Data['Profession'] = Data.Profession.map(Profession_map_dict)
Data
```

```python
# Example-1 :
Data_1 = Data
Data_1

# Here, Data_1 and Data are just two names for same data
```

```python
# Example-2 : Deep Copy

Fav_Places = pd.DataFrame({
                            'Place' : ['England','Newzealand','Switzerland','Antarctica'],
                            'Package' : [80,60,40,50]
})

Fav_Places
```

```python
New_Copy = Fav_Places.copy()

New_Copy.iloc[3,0] = 'Ireland'
New_Copy
```

```python
Fav_Places
```

```python
# Example-3 : Shallow Copy --> deep = False

# New Data Structure but Common shared Data

Fav_Places_2 = pd.DataFrame({
                            'Place' : ['London','Queenstown','Adelaide','Auckland'],
                            'Package' : [80,60,40,50]
})

New_Copy_2 = Fav_Places_2.copy(deep = False)

# New_Copy_2.iloc[3,0] = 'Dublin'
Fav_Places_2
```
