# Creating New Column in DataFrame

```python
import pandas as pd
```

```python
Data = pd.read_excel("Excel_Data.xlsx")

Data['Rank'] = Data.salary.rank()
Data

# Here, rank() function allots the ranking based on the salary
# Lower the Salary --> Lower the Rank, Higher the Salary --> Higher the Rank

Data['Bonus'] = 100000
Data

# Create Age_Rank Column which store rank based on Age

Data['Age_Rank'] = Data.Age.rank()
Data
```

```python
# Sorting the DataFrame based on Particular Column

Data.sort_values('Rank')
```

# Delete the Column from DataFrame

```python
# drop() Function is used to delete the column

Data = Data.drop('Rank', axis = 1)
Data

# axis = 1 --> to inform the drop() function that we want to delete column
```

```python
Data = Data.drop('Age_Rank', axis = 1)
Data
```

```python
# We Can also delete Multiple columns

Data = Data.drop(['Marriage_proposal_count','Bonus'], axis = 1)
Data
```

```python
Data['Bonus','Experience'] = 10
Data['Bonus'] = 100
Data
```

```python
Data = Data.drop([('Bonus','Experience'),'Bonus'], axis = 1)
Data
```

# Replacing Values in DataFrame

```python
Data.Name = Data.Name.replace('olivia','Knight')
Data

Data.Name = Data.Name.replace('Rocky','Brook')
Data
```

```python
# Replacing Multiple values Together

Data.Name = Data.Name.replace({"jake" : "Root", "John" : "Archer"})
Data
```

```python
# Replace Multiple Values in Multiple Columns

Data = Data.replace({"Knight" : "Vince", 23.0 : 36.0, "Teacher" : "Cricketer"})
Data
```

```python
# To Know Data Types of Columns

Data.dtypes
```

```python
Data.Name = Data.Name.astype('string')
Data.dtypes

# Converting salary from int to float

Data.salary = Data.salary.astype('float')
Data.dtypes
```
