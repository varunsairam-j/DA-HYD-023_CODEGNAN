# Handling Missing Values (Null Values)

```python
import pandas as pd
```

```python
Data = pd.read_excel("Excel_Data.xlsx")
Data
```

```python
# Null Values is a Missing value, which is not provided by the User
# Zero is not a Null Value.

Data.isnull() 
```

```python
Data.isnull().sum()     # finds the total number of null values in each column
```

```python
# Print the Record of the People whose Age is Null

Data[Data.Age.isnull()]

# NaN / Na  -->  Null Values
```

```python
# Delete the Null Values

Data_1 = Data.copy()     # copy() function will create a duplicate copy

Data_1 = Data_1.dropna()     # dropna() deletes Null Values
Data_1
```

```python
Data_1.shape
```

```python
Data.shape
```

```python
Data_1.Age.isnull().sum()
```

```python
# Delete the Column which is having Null Values (Age column)

Data_2 = Data.copy()

Data_2 = Data_2.dropna(axis = 1)     # axis = 1 --> deletes the column. By Default, axis = 0 for rows
Data_2
```

```python
# Fill the Null values with mean/average

Data_3 = Data.copy()
Avg_Age = Data_3.Age.mean()     # mean() function calculates Average (Mean)
Data_3 = Data.Age.fillna(Avg_Age)     # fillna() function insert data where the values are Null
```

```python
Data_3.isnull().sum()
```

```python
Data_4 = Data.copy()

Data_4 = Data_4.fillna(0)
Data_4.isnull().sum()
```

# Handling Duplicate Values

```python
# Duplicate record is a Record where every values of a row matching with other row

Data.duplicated().sum()
```

```python
Data[Data.duplicated()]
```

```python
Data[Data.duplicated(keep = False)]
```

```python
# Delete the Duplicated Records

Data_x = Data.copy()

Data_x = Data_x.drop_duplicates()
Data_x
```

# Handling the Inconsistent Data

```python
Data.columns
```

```python
# We find a Inconsistency in Column Names, All Column names are starting with Capital except 'salary'

Data.columns = Data.columns.str.title()
Data.columns
```

```python
# Query : Find the unique Professions

Data.Profession.unique()

# Inconsistency in Profession names
```

```python
# we have to remove spaces before 'Boxer'

Data.Profession = Data.Profession.str.strip()
Data.Profession.unique()
```

```python
Data.Name.unique()
```

```python
Data.Name = Data.Name.str.title()
Data.Name = Data.Name.str.strip()
Data.Name.unique()

# We Can write both of them in a single line

Data.Name = Data.Name.str.title().str.strip()
Data.Name.unique()
```

# Data Transformation using apply(), map()

```python
# apply() : Helps to perform small operation on the DataFrame
# apply() function use the "Lambda" function to perform small operation

Data_9 = Data.copy()

# Query : Sai Ram says that he will doub;e the salary this diwali.

Data_9.Salary = Data_9.Salary.apply(lambda x : x * 2)
Data_9
```

```python
Data_9.Marriage_Proposal_Count = Data_9.Marriage_Proposal_Count.apply(lambda x : x + 5)
Data_9
```
