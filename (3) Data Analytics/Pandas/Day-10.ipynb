# Time Series and Data Generation

```python
# If any column in the Dataset is related to Date or Time, that Dataset is kniwn as Time Series Dataset.

import pandas as pd
```

```python
# date_range()  :  Used for generating Date/Time related data

Time_Series_Data = pd.date_range(start = "2020-01-01", end = "2026-01-01", freq = 'YE')     # YE - Year End
Time_Series_Data 
```

```python
Time_Series_Data_1 = pd.date_range(start = "2020-01-01", end = "2026-01-01", freq = 'YS')     # YS - Year Start
Time_Series_Data_1
```

```python
Time_Series_Data_2 = pd.date_range(start = "2020-01-01", end = "2026-01-01", freq = 'MS')     # MS - Month Start
Time_Series_Data_2
```

```python
Time_Series_Data_3 = pd.date_range(start = "2020-01-01", end = "2026-12-31", freq = 'ME')     # ME - Month End
Time_Series_Data_3
```

```python
Time_Series_Data_4 = pd.date_range(start = "2020-01-01", end = "2026-12-31", freq = 'D')     # D - Day
Time_Series_Data_4
```

```python
Time_Series_Data_5 = pd.date_range(start = "2020-01-01", end = "2026-01-01", freq = 'h')     # h - Hour
Time_Series_Data_5
```

```python
# Select the Random Date/Time and build the DataFrame

import numpy as np
```

```python
Random_Dates = np.random.choice(Time_Series_Data_5, size = 1000)
Random_Dates

# choice() function is helping to select the random 1000 data values
```

```python
len(Random_Dates)
```

```python
type(Random_Dates)
```

```python
# Create DataFrame with Date/Time Column

Df = pd.DataFrame( 
    { 'Time_Stamp' : pd.to_datetime(Random_Dates),
      'Order_id' : np.arange(1,1001),
      'Food_items' : np.random.choice(['Biryani','Dosa','Burger','Samosa','Sandwich'], size = 1000) })

Df

# pd.to_datetime()  :  Used to Convert numpy date/time data into Pandas date/time format

# np.arange()(1,1001)  :  Used to Generate Values from 1 to 1000
```

```python
# np.random.seed(42)
Df_1 = pd.DataFrame( 
    { 'Time_Stamp' : pd.to_datetime(Random_Dates),
      'Order_id' : np.arange(1,1001),
      'Food_items' : np.random.choice(['Biryani','Dosa','Burger','Samosa','Sandwich'], size = 1000) })

Df_1
```

```python
Df_2 = pd.DataFrame( 
    { 'Time_Stamp' : pd.to_datetime(Random_Dates),
      'Order_id' : np.arange(1,1001),
      'Food_items' : np.random.choice(['Biryani','Dosa','Burger','Samosa','Sandwich'], size = 1000) })

Df_2
```

```python
Df['Year'] = Df['Time_Stamp'].dt.year
Df
```

```python
Df['Month'] = Df['Time_Stamp'].dt.month
Df
```

```python
Df['Day'] = Df['Time_Stamp'].dt.day
Df
```

```python
Df['Hour'] = Df['Time_Stamp'].dt.hour
Df['Minute'] = Df['Time_Stamp'].dt.minute
Df['Second'] = Df['Time_Stamp'].dt.second
Df
```

# Data Visualization

```python
# Pandas can also be used to plot simple charts

Df.groupby('Year')['Order_id'].sum().plot(kind = 'bar')

#plot() function is used for plotting the charts in the pandas
```
