```python
# uniform()  :  Used for generating Uniform distributed data

# Unform distributed data --> The Data in which all values are Equally likely.

Uniform_Data = np.random.uniform(low = 5, high = 15, size = 100)
Uniform_Data
```

```python
# Seaborn is used for Plotting

import seaborn as sb

sb.histplot(Uniform_Data, bins = 20)
```

# Strandard Normal Distributed Data

```python
# Strandard Normal distributed data  :  Bell shaped Curve

# randn() --> Used to generate Strandard Normal distributed data

T = np.random.randn(100)
T
```

```python
sb.histplot(T, bins = 30, kde = True)     # kde = True draws a line (Curve) on the Chart
```

```python
# 2D Strandard Normal Distributed data

F = np.random.randn(2,4)
F
```

# Normal Distributed Data

```python
# Generating Normal Distributed Data

# normal()  :  Used to generate Normal Distributes data

Normal_Data = np.random.normal(loc = 5, scale = 2, size = 1000)
Normal_Data
```

```python
sb.histplot(Normal_Data, bins = 50, kde = True)
```
