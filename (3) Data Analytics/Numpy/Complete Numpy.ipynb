# Insertion and Deletion on Numpy array

```python
import numpy as np
```

```python
# Inserting a New Row

Teams = np.array([['Manchester Originals','Northern Superchargers','Oval Invincibles'],
                  ['Trent Rockets','Welsh Fire','Birmingham Pheonix']])
print(Teams)

Updated_Teams_Array = np.insert(Teams,1,["Southern Brave","London Spirit","Surrey"],axis=0)
Updated_Teams_Array

# Here, 1 is showing the index at which we are inserting the Row
# axis = 0 means, We are Inserting the Row
# axis = 1 means, We are Insering the Column
```

```python
Updated_Teams_Array_2 = np.insert(Updated_Teams_Array,1,["Somerset","Middlesex","Sussex"],axis=1)
Updated_Teams_Array_2
```

```python
Updated_Teams_Array_3 = np.insert(Updated_Teams_Array_2,4,["Lancashire","Yorkshire","Hampshire"],axis = 1)
Updated_Teams_Array_3
```

```python
# Deleting a Row

Deleted_Teams_Array = np.delete(Updated_Teams_Array_3,1,axis = 0)
Deleted_Teams_Array
```

```python
# Deleting a Column

Deleted_Teams_Array_2 = np.delete(Deleted_Teams_Array,2, axis=1)
Deleted_Teams_Array_2
```

# Splitting Numpy array

```python
import numpy as np

# Split the 1D Array

Speed = np.array([40,60,70,50,80,100])
Speed
```

```python
# split()  :  Used for splitting the numpy array

Split_Speed = np.split(Speed,2)
Split_Speed

# split() function return a list of sub arrays
```

```python
Split_Speed_2 = np.split(Speed,3)
Split_Speed_2

# split() function can divide an array only if each sub array get equal number of elements
```

```python
Split_Speed_2[0][0]
```

```python
print(Split_Speed_2[2][0])
```

```python
# split 2D array

Bike_Speed = np.array([[60,70,80],
                       [90,100,120]])
Bike_Speed
```

```python
# vsplit()  :  splitting the array Row wise

Vertical_Split = np.vsplit(Bike_Speed,2)
Vertical_Split
```

```python
Bike_Speed_2 = np.array([[60,70,80],
                        [90,100,120],
                        [40,50,60],
                        [110,130,150]])
V_Split_2 = np.vsplit(Bike_Speed_2,2)
V_Split_2
```

```python
V_Split_2[0][1][2]
```

```python
# Horizontal Splitting
# hsplit()  :  Used to split column wise

Car_Speed = np.array([[160,170,180],
                      [190,100,120],
                      [140,150,160],
                      [110,130,150]])

H_Split = np.hsplit(Car_Speed,3)
H_Split
```

```python
H_Split[1][2][0]
```

```python
H_Split[-2][-2][-1]
```
