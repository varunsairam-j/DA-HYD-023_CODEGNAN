# Indexing, Slicing and Modification

```python
Movies = np.array([["The Prestige","Jersey","K.G.F"],
                  ["Avengers","Man of Steel","Justice League"]])
print(Movies)
Movies[0,1]     # Accessing Jersey
```

```python
# Accessing Avengers

Movies[1,0]     # [Row,Column]
    #or
Movies[1][0]
```

```python
# Negative Indexing

Movies[-2,0]
    #or
Movies[-2][0]
```

```python
Movies[-1][-1]
```

```python
Movies[1]
   #or
Movies[1,:]     # [Row,Column]
```

```python
Movies[:,2]
```

```python
Movies[:,:]
```

```python
X = Movies[0,0],Movies[1,1]
print(X)
```

```python
Movies[-1:, -1:-2:-1]
```

# Modification on Numpy array

```python
Players = np.array([['Tom Banton','Ben Duckett','Joe Root'],
                    ['Harry Brook','Jordan Cox','Will Jacks'],
                    ['Sam Curran','Gus Atkinson','Liam Dawson'],
                    ['Jofra Archer','Sonny Baker','Adil Rashid']])
# Modification
Players[1,1] = "Jos Buttler"
Players
```

```python
Players[0,1] = "Jacob Bethell"     # Here, When Modification, It will store only upto the places of previous elements
Players

# As 'Ben Duckett' (11 characters) and 'Jacob Bethell' (13 characters), We Cannot fit the Full name 
```

```python
# Modification of Complete Row

Players[0] = ['Jason Roy','Johnny Bairstow','Eion Morgan']
Players
```

```python
# Modification of Complete Column

Players[ :,0] = ['Alex Hales','Harry Brook','Ben Stokes','Moeen Ali']
Players
```
