# More Statistical Operations

```python
import numpy as np

Reel_Count = np.array([1000,500,600,300,1250,700,250,400])
Reel_Count
```

```python
# argmax()  :  Returns the Index of maximum Element

print(np.argmax(Reel_Count))
```

```python
X = np.array([[20,24,36,28],
              [12,40,16,26]])

print(np.argmax(X))
```

```python
# argmin()  :  Returns the Index of Minimum Value

print(np.argmin(X))
```

```python
print(np.log10(100))
```

```python
print(np.log2(1024))
```

```python
print(np.log(10))

# By Default, log base is e = 2.71
```

```python
# sqrt()  :  Taking Square Root of a Number 

print(np.sqrt(400))
```

```python
# round()  :  Rounds the Value upto given decimal Places

Pi = 3.14159265359
print(np.round(Pi,2))
```

```python
print(np.round(Pi))
```

```python
# floor()  :  Rounds the Value to Lower Bound

print(np.floor(Pi))
```

```python
# ceil()  :  Rounds the Value to Upper Bound

print(np.ceil(Pi))
```

```python
# abs()  :  Converts Negative to Positive

Value = -32
print(np.abs(Value))
```

```python
# sort()  :  Sorts the Elemeents in Lexographical Order

# Lexographical Order  :  If First Letter is matched. Then, Compare the Next Character

England = np.array(['Ben Duckett', 'Tom Banton', 'Joe Root', 'Harry Brook', 'Jordan Cox', 'Will Jacks', 'Sam Curran',
                    'Liam Dawson', 'Gus Atkinson', 'Jofra Archer', 'Sonny Baker'])
np.sort(England)
```

```python
# Query :  Sort the this array

Reel_Count = np.array([1000,500,600,300,1250,700,250,400])
np.sort(Reel_Count)
```

```python
P = np.array([[1,4,3,7,2],
              [7,6,9,2,1],
              [3,9,4,6,2]])
np.sort(P)
```

```python
# Reverse Sort

np.sort(Reel_Count)[::-1]
```

```python
np.sort(P)[::-1]
```

# Concatenation, Stacking & Splitting of numpy arrays

```python
# concatenate()  :  This Function will put content of two or more numpy array's into single array

A = ['Varun','Sai','Ram']
B = ['Sir','James','Anderson']

X = np.concatenate((A,B))

print(X)
```

```python
print(A+B)
```

```python
C = ['Benjamin','Andrew','Stokes']
D = ["Marvel's", "The Avengers"]

Z = np.concatenate((A,B,C,D))
print(Z)
```

```python
# Concatenating 2D Arrays  :  Horizontal Concatenation (Column wise / Side by Side)

Boys_Dream_Package = np.array([[20,40,60],
                               [10,30,50]])

Girls_Dream_Package = np.array([[15,25,35],
                                [5,20,30]])

New_Array = np.concatenate((Boys_Dream_Package, Girls_Dream_Package), axis = 1)
New_Array
```

```python
New_Array_2 = np.concatenate((Boys_Dream_Package, Girls_Dream_Package), axis = 0)
New_Array_2
```

```python
New_Array_3 = np.concatenate((Boys_Dream_Package, Girls_Dream_Package))
New_Array_3

# By Default, axis = 0
```

```python
# stack()  :  It merges arrays with new dimension

Array_1 = np.array([10,20,30])
Array_2 = np.array([40,50,60])

Stacked_Array = np.stack((Array_1,Array_2), axis = 0)
Stacked_Array
```

```python
Concatenated_Array = np.concatenate((Array_1,Array_2), axis = 0)
Concatenated_Array
```
