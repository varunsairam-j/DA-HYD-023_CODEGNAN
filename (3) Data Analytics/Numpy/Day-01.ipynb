# Special type of Numpy array creation

```python
# Array with Zero value initialization

A = np.zeros(5)
A
```

```python
B = np.zeros(10)
B
```

```python
# 2D Zeros array

C = np.zeros((2,5))
C
```

```python
# 3D array of zeros with dimensions (2,3,5)

D = np.zeros((2,3,5))
D
```

```python
# 4D array of zeros

E = np.zeros((2,2,2,4))
E
```

```python
# 5D

F = np.zeros((2,2,2,3,5))
F
```

```python
# 1D array of Ones

X = np.ones(5)
X
```

```python
# 2D array of Ones

Y = np.ones((3,5))
Y
```

```python
# 3D array of Ones

Z = np.ones((2,3,5))
Z
```

```python
# full()  :  Used to Initialize with any Number

V = np.full(5,10)
V
```

```python
# 2D array initialized with 9

ABC = np.full((2,5),9)
ABC
```

```python
# 3D array initialized with 7

ABCD = np.full((3,2,4),7)
ABCD
```
