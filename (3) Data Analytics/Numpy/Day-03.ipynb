# arange(), shuffle(), permutation(), linspace()

```python
# arange(start, end, step)  :  Generate sequence values from start to end-1 with given step size

Array = np.arange(1,10)
Array

# Similar to range() in python
```

```python
Array_2 = np.arange(1,10,2)
Array_2
```

```python
Array_3 = np.arange(2,100,4)
Array_3
```

```python
# shuffle()  :  this will shuffle the Elements

np.random.shuffle(Array_3)
Array_3
```

```python
# permutation()  :  this will ra-arrange the elements

Array_4 = np.random.permutation(Array_3)
Array_4
```

```python
np.random.shuffle(Array_4)
Array_4
```

```python
# linspace()  :  linear space function generate the same difference elements

Z = np.linspace(1,10,5)
Z
```

```python
W = np.linspace(10,100,25)
W
```

# Array Size & Re-shaping & Transpose

```python
South_Foods = np.array([['Dosa','Idly','Upma','Bonda'],
                       ['Rice','Pachadi','Sambar','Curd']])
South_Foods
```

```python
South_Foods.shape
```

```python
# reshape()  :  used to convert dimension of array

Reshaped_Array = South_Foods.reshape(8)
Reshaped_Array

# Reshaped the array to 1D
```

```python
# Reshaping to (4,2) from (2,4)

Reshaped_Array_2 = South_Foods.reshape(4,2)
Reshaped_Array_2
```

```python
# Reshaping to 3D

Reshaped_Array_3 = South_Foods.reshape(2,2,2)
Reshaped_Array_3
```

```python
# Reshaping to 4D

Reshaped_Array_4 = South_Foods.reshape(1,2,2,2)
Reshaped_Array_4
```

```python
Reshaped_Array_4.ndim
```

```python
# ravel()  :  Convert any D array to 1D array

New_Array = South_Foods.ravel()
New_Array
```

```python
# transpose()  :  Rows <-> Columns

South_Foods
```
