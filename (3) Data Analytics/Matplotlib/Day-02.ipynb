# Line Chart

```python
import pandas as pd

Dates = pd.date_range(start = "2026-01-01", periods = 12, freq = 'ME')
Dates
```

```python
Sales = [100,80,130,90,120,100,110,140,130,120,90,100]
```

```python
# plot()  :  Used to plot the line chart

plt.figure(figsize = (6,4))
plt.plot(Dates, Sales)
```

```python
plt.figure(figsize = (6,4))
plt.plot(Dates, Sales, marker = 'o',color = 'm')

# xticks()  : To tilt the values on X-axis with specified angle

plt.xticks(Dates,rotation = 60)          

plt.show()         # To Show Only needed Information (Hiding the Matter)
```

```python
plt.figure(figsize = (6,4))
plt.plot(Dates, Sales, marker = 'o',color = 'g')

plt.xticks(Dates,Dates.strftime('%B'),rotation = 60)     

# '%b' : Month Name in short
# '%B' : Month Name in Full
# '%Y' : Year

plt.title('Sales in 2026')
plt.xlabel('Months')
plt.ylabel('Sales')

plt.grid(True)      # gives Grid lines to the Chart

plt.show()
```

# Multiline Chart

```python
Dates = pd.date_range(start = '2005-01-01', periods = 15, freq = 'YE')
Dates
```

```python
Bikes = [50,70,60,90,100,80,130,150,200,300,180,220,160,290,270]
Cars = [10,30,50,70,120,160,100,120,180,240,160,260,130,250,230]
Cycles = [150,170,160,190,150,180,170,90,80,120,100,120,100,90,50]

plt.figure(figsize = (6,3))
plt.plot(Dates, Bikes, color = 'b', marker = 'o', label = 'Bikes')          # label -> Used to give Labels to the lines
plt.plot(Dates,Cars, color = 'g', marker = 'x', label = 'Cars')
plt.plot(Dates,Cycles, color = 'orange', marker = 'v', label = 'Cycles')

# legend()  :  Provides the details of different lines
# legend()  :  labels different catrgory of data

plt.legend()
```

# Scatterplot

```python
# Positive Relation

Age = [10,12,14,22,28,16,24,30,20,25]
Salary = [15,20,25,40,50,22,55,25,30,36]

plt.figure(figsize = (4,4))

plt.scatter(Age, Salary, color = 'c')

plt.xlabel("Age")
plt.ylabel("Salary")
plt.title("Relation between Age & Salary")
```

```python
# Negative Relation

Age = [10,15,20,25,30,35,40,45,50,55,60]
Time_to_Die = [70,65,60,55,50,45,40,35,30,25,20]

plt.figure(figsize = (4,3))
plt.scatter(Age,Time_to_Die, color = 'y')

plt.title("Age & Time to Die")
plt.xlabel("Age")
plt.ylabel("No.of years more to die")
```
