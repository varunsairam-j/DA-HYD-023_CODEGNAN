# Install and Import matplotlib

```python
# pip install matplotlib

import matplotlib.pyplot as plt

import matplotlib as mpl

mpl.__version__
```

# Bar Chart

```python
Breakfast_Items = ['Pongal','Idly','Dosa','Puri','Bonda','Vada']
No_of_Plates = [1,6,5,7,10,6]

plt.figure(figsize = (5,3))     # plt.figure is used for adjusting the size of the chart
plt.bar(Breakfast_Items, No_of_Plates)     # plt.bar(x-axis,y-axis)
```

```python
# We Can give customized colors to our charts

plt.figure(figsize = (5,3))
plt.bar(Breakfast_Items, No_of_Plates, color = "Green")
```

```python
# We Can also give Individual color to Individual bar.

plt.figure(figsize = (6,3))
plt.bar(Breakfast_Items, No_of_Plates, color = ["g","m","y","b","c","r"])
```

```python
# we can also give text/value over the bars

# plt.text(x,y)  :  Used to add the text/value in the chart

plt.figure(figsize = (4,3))
plt.bar(Breakfast_Items, No_of_Plates, color = ["g","m","y","b","c","r"])

plt.text(0,2,"2")
plt.text(1,7,"6")
plt.text(2,6,"5")
plt.text(3,8,"7")
plt.text(4,10,"10")
plt.text(5,7,"6")
```

```python
# using for loop to give text to bars

Breakfast_Items = ['Pongal','Idly','Dosa','Puri','Bonda','Vada']
No_of_Plates = [1,6,5,7,8,6]

plt.figure(figsize = (5,3))     # plt.figure is used for adjusting the size of the chart
plt.bar(Breakfast_Items, No_of_Plates,color = ["g","m","y","b","c","r"])     # plt.bar(x-axis,y-axis)

for x,y in enumerate(No_of_Plates):     # enumerate() is used to generate the index.
    plt.text(x,y+1,y)

plt.title("Food & No.of Plates")
plt.ylabel("No.of Plates")
plt.xlabel("Food Items")
```

# Horizontal Bar Chart

```python
# Def  :  Categorical Data is plotted on Y-axis
# Big Names are handled without any problem in Horizontal Bar chart without Overlapping with other names.

Breakfast_Items = ['Pongal','Idly','Dosa','Puri','Bonda','Vada']
No_of_Plates = [1,6,5,7,10,6]

plt.barh(Breakfast_Items, No_of_Plates, color = ['g','r','c','y','m','b'])

plt.title("Food & No.of Plates")
plt.xlabel("No.of Plates")
plt.ylabel("Food Item")
```

# Histogram

```python
import numpy as np           # to generate Random Values of Normal distributed data

Data = np.random.randn(1000)     # Generates Normal distributed data

plt.figure(figsize = (4,3))
plt.hist(Data, bins = 20, edgecolor = 'black')

plt.title("Normal Distribution")
plt.xlabel("Continous Data")
plt.ylabel("Total Value = 1000")

plt.show()        # Removes Unwanted data
```
