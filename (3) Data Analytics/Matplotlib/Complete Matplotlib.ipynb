# Bar Chart

```python
Breakfast_Items = ['Pongal','Idly','Dosa','Puri','Bonda','Vada']
No_of_Plates = [1,6,5,7,10,6]

plt.figure(figsize = (5,3))     # plt.figure is used for adjusting the size of the chart
plt.bar(Breakfast_Items, No_of_Plates)     # plt.bar(x-axis,y-axis)
```

```python
# We Can give customized colors to our charts

plt.figure(figsize = (5,3))
plt.bar(Breakfast_Items, No_of_Plates, color = "Green")
```

```python
# We Can also give Individual color to Individual bar.

plt.figure(figsize = (6,3))
plt.bar(Breakfast_Items, No_of_Plates, color = ["g","m","y","b","c","r"])
```

```python
# we can also give text/value over the bars

# plt.text(x,y)  :  Used to add the text/value in the chart

plt.figure(figsize = (4,3))
plt.bar(Breakfast_Items, No_of_Plates, color = ["g","m","y","b","c","r"])

plt.text(0,2,"2")
plt.text(1,7,"6")
plt.text(2,6,"5")
plt.text(3,8,"7")
plt.text(4,10,"10")
plt.text(5,7,"6")
```

```python
# using for loop to give text to bars

Breakfast_Items = ['Pongal','Idly','Dosa','Puri','Bonda','Vada']
No_of_Plates = [1,6,5,7,8,6]

plt.figure(figsize = (5,3))     # plt.figure is used for adjusting the size of the chart
plt.bar(Breakfast_Items, No_of_Plates,color = ["g","m","y","b","c","r"])     # plt.bar(x-axis,y-axis)

for x,y in enumerate(No_of_Plates):     # enumerate() is used to generate the index.
    plt.text(x,y+1,y)

plt.title("Food & No.of Plates")
plt.ylabel("No.of Plates")
plt.xlabel("Food Items")
```

# Horizontal Bar Chart

```python
# Def  :  Categorical Data is plotted on Y-axis
# Big Names are handled without any problem in Horizontal Bar chart without Overlapping with other names.

Breakfast_Items = ['Pongal','Idly','Dosa','Puri','Bonda','Vada']
No_of_Plates = [1,6,5,7,10,6]

plt.barh(Breakfast_Items, No_of_Plates, color = ['g','r','c','y','m','b'])

plt.title("Food & No.of Plates")
plt.xlabel("No.of Plates")
plt.ylabel("Food Item")
```

# Histogram

```python
import numpy as np           # to generate Random Values of Normal distributed data

Data = np.random.randn(1000)     # Generates Normal distributed data

plt.figure(figsize = (4,3))
plt.hist(Data, bins = 20, edgecolor = 'black')

plt.title("Normal Distribution")
plt.xlabel("Continous Data")
plt.ylabel("Total Value = 1000")

plt.show()        # Removes Unwanted data
```

# Line Chart

```python
import pandas as pd

Dates = pd.date_range(start = "2026-01-01", periods = 12, freq = 'ME')
Dates
```

```python
Sales = [100,80,130,90,120,100,110,140,130,120,90,100]
```

```python
# plot()  :  Used to plot the line chart

plt.figure(figsize = (6,4))
plt.plot(Dates, Sales)
```

```python
plt.figure(figsize = (6,4))
plt.plot(Dates, Sales, marker = 'o',color = 'm')

# xticks()  : To tilt the values on X-axis with specified angle

plt.xticks(Dates,rotation = 60)          

plt.show()         # To Show Only needed Information (Hiding the Matter)
```

```python
plt.figure(figsize = (6,4))
plt.plot(Dates, Sales, marker = 'o',color = 'g')

plt.xticks(Dates,Dates.strftime('%B'),rotation = 60)     

# '%b' : Month Name in short
# '%B' : Month Name in Full
# '%Y' : Year

plt.title('Sales in 2026')
plt.xlabel('Months')
plt.ylabel('Sales')

plt.grid(True)      # gives Grid lines to the Chart

plt.show()
```

# Multiplots and Subplots

### Method-1 : plt.subplot()

```python
# Syntax :
    # --> plt.subplot(nrows, ncols, index)

# nrows  :  Number of Rows in the Grid of Subplots
# ncols  :  Number of Columns in the grid of subplots
# index  :  subplot number (starts with 1)

import matplotlib.pyplot as plt
```

```python
Runs_type = ['Singles', 'Doubles', 'Triples', 'Fours', 'Sixes']
Root = [20,15,5,35,25]
Duckett = [15,10,5,50,20]
Brook = [5,15,5,40,40]
Banton = [20,10,5,40,25]
Bethell = [20,10,10,40,20]
Buttler = [15,5,5,40,35]

plt.figure(figsize = (12,8))

plt.subplot(2,3,1)
plt.pie(Root, labels = Runs_type, autopct = '%1.1f%%')
plt.title("Root")

plt.subplot(2,3,2)
plt.bar(Runs_type, Duckett, color = ['m','c','g','orange','b','r'])
plt.title("Duckett")
    
plt.subplot(2,3,3)
plt.barh(Runs_type, Brook, color = ['c','g','orange','m','b','r'])
plt.title("Brook")

plt.subplot(2,3,4)
plt.pie(Banton, labels = Runs_type, autopct = '%1.1f%%')
plt.title("Banton")

plt.subplot(2,3,5)
plt.pie(Bethell, labels = Runs_type, autopct = '%1.1f%%')
plt.title("Bethell")

plt.subplot(2,3,6)
plt.pie(Buttler, labels = Runs_type, autopct = '%1.1f%%')
plt.title("Buttler")

plt.tight_layout()

plt.show()
```

### Method-2 : plt.subplots()

```python
Player = ['Salt','Brook','Buttler','Jacks','Curran','Bethell']
Runs = [30,70,50,25,20,40]
```

```python
fig,axes = plt.subplots(2,2, figsize = (8,4))

# fig = the whole figure/canvas
# figsize = Size of the Complete Canvas
# axes = Array of Subplots (2 x 2)

axes     # 2D Array

One_D_Array = axes.ravel()     # Converting 2D Array to 1D Array
One_D_Array

for ax in One_D_Array:
    ax.pie(Runs, labels = Player, counterclock = False)

plt.tight_layout()
plt.show()
```

```python
fig,axes = plt.subplots(2,2, figsize = (8,4))

# fig = the whole figure/canvas
# figsize = Size of the Complete Canvas
# axes = Array of Subplots (2 x 2)

axes     # 2D Array

One_D_Array = axes.ravel()     # Converting 2D Array to 1D Array
One_D_Array

labels = ['Pie-1','Pie-2','Pie-3','Pie-4']

for ax,label in zip(One_D_Array, labels):
    ax.pie(Runs, labels = Player, counterclock = False)
    ax.set_title(label)

# set_title()  :  Used to set the title of subplots
# zip()  :  To Access the data element by element from multiple sub arrays

plt.tight_layout()
plt.show()
```

# 3D Chart using Matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# plt.figure() is used to set the size of the Image Frame

Figure = plt.figure()

# add_plot() have 3d Projection Parameter, which helps us to chart 3d chart

A = Figure.add_subplot(111, projection = '3d')        # 111 --> Row Column Index

X,Y,Z = np.random.rand(3,100)     # Generating 100 random elements for each dimension/varaiable x,y,z

A.scatter(X,Y,Z)

plt.show()
```
