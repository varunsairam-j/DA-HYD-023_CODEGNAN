# Pie Chart

```python
import matplotlib.pyplot as plt
```

```python
Mobiles = ['OnePlus','IPhone','Samsung','Vivo','Oppo','Others']
Shares = [15,20,35,10,15,5]

plt.figure(figsize = (5,4))
plt.pie(Shares, labels = Mobiles, autopct = '%1.1f%%', counterclock = False)

plt.title("Mobile Market Shares")

# savefig()  :  To Save Chart into Current Working Directory (Our System)

plt.savefig("Mobile Shares Pie_Chart.jpg")
plt.savefig("Mobile Shares Pie_Chart.pdf")
plt.savefig("Mobile Shares Pie_Chart.png")
```

# Donut

```python
Teams = ['CSK','MI','RCB','SRH','KKR','Others']
Fans = [25,20,25,15,10,5]

plt.figure(figsize = (5,4))

plt.pie(Fans, labels = Teams, autopct = '%1.1f%%', wedgeprops = dict(width = 0.5), counterclock = False)

plt.title("IPL Teams Fans %")

plt.text(0,0, "Fans %", ha = 'center')
```

# Box Plot

```python
Salary = [-10,42,30,35,34,20,40,45,50,80,90]

plt.figure(figsize = (5,4))
plt.boxplot(Salary)

plt.show()

# Upper Bound, Lower Bound
# Q1 --> Min (The Line below the Box)
# Q2 --> Median (The Yellow Line in the Box)
# Q3 --> Max (The Line above the Box)
# Outliers --> Cicles away from the box
# Whisker --> The Line
```
