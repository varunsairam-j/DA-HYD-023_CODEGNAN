# Multiplots and Subplots

### Method-1 --> plt.subplot()

```python
# Syntax :
    # --> plt.subplot(nrows, ncols, index)

# nrows  :  Number of Rows in the Grid of Subplots
# ncols  :  Number of Columns in the grid of subplots
# index  :  subplot number (starts with 1)

import matplotlib.pyplot as plt
```

```python
Runs_type = ['Singles', 'Doubles', 'Triples', 'Fours', 'Sixes']
Root = [20,15,5,35,25]
Duckett = [15,10,5,50,20]
Brook = [5,15,5,40,40]
Banton = [20,10,5,40,25]
Bethell = [20,10,10,40,20]
Buttler = [15,5,5,40,35]

plt.figure(figsize = (12,8))

plt.subplot(2,3,1)
plt.pie(Root, labels = Runs_type, autopct = '%1.1f%%')
plt.title("Root")

plt.subplot(2,3,2)
plt.bar(Runs_type, Duckett, color = ['m','c','g','orange','b','r'])
plt.title("Duckett")
    
plt.subplot(2,3,3)
plt.barh(Runs_type, Brook, color = ['c','g','orange','m','b','r'])
plt.title("Brook")

plt.subplot(2,3,4)
plt.pie(Banton, labels = Runs_type, autopct = '%1.1f%%')
plt.title("Banton")

plt.subplot(2,3,5)
plt.pie(Bethell, labels = Runs_type, autopct = '%1.1f%%')
plt.title("Bethell")

plt.subplot(2,3,6)
plt.pie(Buttler, labels = Runs_type, autopct = '%1.1f%%')
plt.title("Buttler")

plt.tight_layout()

plt.show()
```

### Method-2 : plt.subplots()

```python
Player = ['Salt','Brook','Buttler','Jacks','Curran','Bethell']
Runs = [30,70,50,25,20,40]
```

```python
fig,axes = plt.subplots(2,2, figsize = (8,4))

# fig = the whole figure/canvas
# figsize = Size of the Complete Canvas
# axes = Array of Subplots (2 x 2)

axes     # 2D Array

One_D_Array = axes.ravel()     # Converting 2D Array to 1D Array
One_D_Array

for ax in One_D_Array:
    ax.pie(Runs, labels = Player, counterclock = False)

plt.tight_layout()
plt.show()
```

```python
fig,axes = plt.subplots(2,2, figsize = (8,4))

# fig = the whole figure/canvas
# figsize = Size of the Complete Canvas
# axes = Array of Subplots (2 x 2)

axes     # 2D Array

One_D_Array = axes.ravel()     # Converting 2D Array to 1D Array
One_D_Array

labels = ['Pie-1','Pie-2','Pie-3','Pie-4']

for ax,label in zip(One_D_Array, labels):
    ax.pie(Runs, labels = Player, counterclock = False)
    ax.set_title(label)

# set_title()  :  Used to set the title of subplots
# zip()  :  To Access the data element by element from multiple sub arrays

plt.tight_layout()
plt.show()
```
